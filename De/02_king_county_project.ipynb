Die Imports bieten drei **Metriken**. Hier ist $y_i$ der echte Preis, $\hat{y}_i$ der vorhergesagte Preis und $n$ die Zahl der Häuser.

$$\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

In Worten: Du quadrierst jeden Fehler, nimmst den Mittelwert und dann die Wurzel.

$$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$

In Worten: Du nimmst den Mittelwert der Fehler ohne ihr Vorzeichen.

$$R^2 = 1 - \frac{\sum_i (y_i - \hat{y}_i)^2}{\sum_i (y_i - \bar{y})^2}$$

In Worten: Das **R²** vergleicht deine quadrierten Fehler mit denen einer Vorhersage, die immer den mittleren Preis $\bar{y}$ nimmt.
