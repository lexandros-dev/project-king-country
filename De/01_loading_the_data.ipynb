## 📖 Neue Begriffe

| Deutsch | Englisch | In einem Satz |
| :--- | :--- | :--- |
| **DataFrame** | DataFrame | Ein **DataFrame** ist eine pandas-Tabelle mit Zeilen und benannten Spalten. |

| **Verknüpfungs-Art** | type of merge | Die **Verknüpfungs-Art** legt fest, welche Zeilen bleiben, wenn ein **Schlüssel** nur in einer Tabelle vorkommt. |

### Die Verknüpfungs-Art wählen

[`pd.merge()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html) hat ein Argument `how`. Es legt die **Verknüpfungs-Art** fest. Sie entscheidet, welche Zeilen bleiben, wenn ein **Schlüssel** nur in einer der zwei Tabellen vorkommt.

| `how` | Behält |
| :--- | :--- |
| `"inner"` | nur die **Schlüssel**, die in beiden Tabellen vorkommen |
| `"left"` | jede Zeile der linken Tabelle `sales`, mit leeren Werten, wo `details` nichts Passendes hat |
| `"right"` | jede Zeile der rechten Tabelle `details`, mit leeren Werten, wo `sales` nichts Passendes hat |
| `"outer"` | jede Zeile beider Tabellen |

Für die Wahl prüfst du, ob die zwei Tabellen vollständig zusammen-passen.
