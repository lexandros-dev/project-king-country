## 📖 New terms

| Term | In one sentence |
| :--- | :--- |
| **DataFrame** | A **DataFrame** is a pandas table with rows and named columns. |
| **key** | A **key** is a column whose values link the rows of two tables. |
| **unique** | A column is **unique** if no value appears in it twice. |
| **one-to-many** | In a **one-to-many** relationship, one row of one table matches several rows of the other table. |
| **merge** | A **merge** combines two tables by matching the values of their **key** columns. |
| **type of merge** | The **type of merge** decides which rows stay when a **key** appears in only one table. |
| **assert** | An **assert** stops the notebook with an error if a condition is false. |

## 3️⃣ Section 3 of 4: Check the key and choose the type of merge

- A **merge** connects each row of `sales` with every matching row of `details`.

- You get one row per sale only if each **key** appears once in `details`.
- If a **key** appeared twice, every sale of that house would be copied.
- The combined table would then have more rows than there are sales.
- pandas shows no error and no warning in this case.

### Choose the type of merge

[`pd.merge()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html) has an argument `how`. It sets the **type of merge**. It decides which rows stay when a **key** appears in only one of the two tables.

| `how` | Keeps |
| :--- | :--- |
| `"inner"` | only the **keys** that appear in both tables |
| `"left"` | every row of the left table `sales`, with empty values where `details` has no match |
| `"right"` | every row of the right table `details`, with empty values where `sales` has no match |
| `"outer"` | every row of both tables |

To choose, find out whether the two tables match completely.

<details>
<summary>💡 Hint</summary>

`.isin()` gives `True` for each value that appears in the other column. `~` turns `True` into `False` and back. `.sum()` counts the `True` values.

</details>

<details>
<summary>✅ Solution</summary>

```python
print("sales without house details:", (~sales["house_id"].isin(details["id"])).sum())
print("houses without a sale:      ", (~details["id"].isin(sales["house_id"])).sum())
```

- Both counts are 0. The two tables match completely.
- So every **type of merge** gives the same result here.
- If a house were missing from `details`, you would choose `"left"`.
- Then every sale stays in the table, because you predict the price of every sale.
- Only the house columns of that sale would be empty.
