## Kapitel 1 von 2: Daten laden und verknüpfen

Lesezeit: ca. 5 Minuten · Fortschritt: ●○

### ⚡ In 30 Sekunden

- Die Daten kommen als zwei Tabellen: 21.597 Verkäufe und 21.420 Häuser.
- Ein Haus kann mehrmals verkauft sein, die Beziehung ist **Eins-zu-viele**.
- Weil der **Schlüssel** `id` in `details` **eindeutig** ist, entsteht genau eine Zeile pro Verkauf.

### 📖 Begriffe

| Deutsch | Englisch | In einem Satz |
| :--- | :--- | :--- |
| **DataFrame** | DataFrame | Ein **DataFrame** ist eine pandas-Tabelle mit Zeilen und benannten Spalten. |
| **Schlüssel** | key | Ein **Schlüssel** ist eine Spalte, deren Werte die Zeilen von zwei Tabellen verbinden. |
| **eindeutig** | unique | Eine Spalte ist **eindeutig**, wenn kein Wert darin zweimal vorkommt. |
| **Eins-zu-viele** | one-to-many | Bei **Eins-zu-viele** passt eine Zeile der einen Tabelle zu mehreren Zeilen der anderen Tabelle. |
| **Verknüpfung** | merge | Eine **Verknüpfung** verbindet zwei Tabellen über gleiche Werte in der Spalte mit dem **Schlüssel**. |
| **Verknüpfungs-Art** | type of merge | Die **Verknüpfungs-Art** legt fest, welche Zeilen bleiben, wenn ein **Schlüssel** nur in einer Tabelle vorkommt. |
| **Assert** | assert | Ein **Assert** stoppt das Notebook mit einem Fehler, wenn eine Bedingung falsch ist. |

### 💻 Wichtiger Code

```python
sales = pd.read_csv("../data/king_county_house_sales.csv")
details = pd.read_csv("../data/king_county_house_details.csv")

# Ist der Schlüssel in details eindeutig?
print("every id is unique:", details["id"].is_unique)

# Fehlen Schlüssel in einer der Tabellen?
print("sales without house details:", (~sales["house_id"].isin(details["id"])).sum())

# Verknüpfung: eine Zeile pro Verkauf
df = sales.merge(details, left_on="house_id", right_on="id",
                 how="left", suffixes=("", "_details"))
assert len(df) == len(sales), "The combined table should have exactly one row per sale"

# Speichern für Notebook 02, ohne Zeilen-Nummern
df.to_csv("../data/eda.csv", index=False)
```

## Kapitel 2.1 von 2: Ziel, Aufteilung, Untersuchung und Basis-Modelle

Lesezeit: ca. 7 Minuten · Fortschritt: ●●

### ⚡ In 30 Sekunden

- Die **Metrik** ist der **RMSE** in Dollar, dazu **MAE** und **R²**.
- Zuerst teilst du in 17.277 Zeilen **Trainings-Daten** und 4.320 Zeilen **Test-Daten**.
- Das **Regel-Basis-Modell** mit 244,44 Dollar pro Quadrat-Fuß ist die Messlatte für echte Modelle.

### 📖 Begriffe

| Deutsch | Englisch | In einem Satz |
| :--- | :--- | :--- |
| **Regression** | regression | Eine **Regression** sagt eine Zahl vorher, hier den Verkaufs-Preis eines Hauses. |
| **Ziel-Variable** | target | Die **Ziel-Variable** ist die Spalte, die du vorhersagst, hier `price`. |
| **Merkmal** | feature | Ein **Merkmal** ist eine Spalte, mit der das Modell die **Ziel-Variable** vorhersagt. |
| **Metrik** | metric | Eine **Metrik** ist eine Zahl, die zeigt, wie gut die Vorhersagen sind. |
| **RMSE** | RMSE | Der **RMSE** ist der typische Vorhersage-Fehler in Dollar, und große Fehler zählen mehr. |
| **MAE** | MAE | Der **MAE** ist der mittlere absolute Vorhersage-Fehler in Dollar. |
| **R²** | R² | Das **R²** ist der Anteil der Preis-Unterschiede, den das Modell erklärt. |
| **Trainings-Daten** | training set | Die **Trainings-Daten** sind der Teil der Daten, den du untersuchst und zum Trainieren nutzt. |
| **Test-Daten** | test set | Die **Test-Daten** sind der gesperrte Teil der Daten für die letzte Bewertung. |
| **Korrelation** | correlation | Die **Korrelation** misst von minus 1 bis plus 1, wie eng zwei Spalten entlang einer Geraden gemeinsam steigen oder fallen. |
| **schief** | skewed | Eine Spalte ist **schief**, wenn wenige sehr große Werte ihre Verteilung zu einer Seite ziehen. |
| **fehlender Wert** | missing value | Ein **fehlender Wert** ist eine leere Zelle, in pandas als `NaN` angezeigt. |
| **Basis-Modell** | baseline | Ein **Basis-Modell** ist ein einfaches Modell, das deine echten Modelle schlagen müssen. |
| **Mittelwert-Basis-Modell** | mean baseline | Das **Mittelwert-Basis-Modell** sagt für jedes Haus den mittleren Preis der **Trainings-Daten** vorher. |
| **Regel-Basis-Modell** | rule-based baseline | Das **Regel-Basis-Modell** sagt den Preis mit einer einfachen Regel aus einer Spalte vorher. |

### 💻 Wichtiger Code

```python
df = pd.read_csv("../data/eda.csv")

# Einmal teilen, fester Zufall, Test-Daten bis zum Schluss sperren
df_train, df_test = train_test_split(df, test_size=0.2, random_state=42)

# Untersuchen nur auf den Trainings-Daten
df_train.corr(numeric_only=True)["price"].sort_values(ascending=False)
df_train.skew(numeric_only=True).sort_values(ascending=False)
df_train.isna().sum()

# Alle Änderungen von Hand in einer Funktion
def prepare(data):
    data = data.copy()
    sale_year = pd.to_datetime(data["date"]).dt.year
    data["age_at_sale"] = sale_year - data["yr_built"]
    data["renovated"] = (data["yr_renovated"] > 0).astype(int)
    return data.drop(columns=["id", "house_id", "id_details", "date",
                              "yr_renovated", "sqft_basement"])

train = prepare(df_train)
X_train, y_train = train.drop(columns="price"), train["price"]

# Zwei Basis-Modelle
mean_baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
price_per_sqft = (y_train / X_train["sqft_living"]).median()
```

$$\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

In Worten: Du quadrierst jeden Fehler, nimmst den Mittelwert und dann die Wurzel.

## Kapitel 2.2 von 2: Pipeline, Abstimmung und Bewertung

Lesezeit: ca. 8 Minuten · Fortschritt: ●●

### ⚡ In 30 Sekunden

- Eine **Pipeline** mit **ColumnTransformer** lernt **Auffüllen**, **Skalieren** und **One-Hot-Kodierung** nur auf den **Trainings-Daten**.
- **Kreuz-Validierung** mit **Zufalls-Suche** und **Gitter-Suche** findet `alpha` für **Ridge** mit und ohne **Polynom-Merkmale**.
- **Ridge** mit **Polynom-Merkmalen** gewinnt mit 123.490 Dollar **RMSE** auf den **Test-Daten**, teure Häuser haben die größten **Residuen**.

### 📖 Begriffe

| Deutsch | Englisch | In einem Satz |
| :--- | :--- | :--- |
| **Pipeline** | pipeline | Eine **Pipeline** verkettet Vorbereitungs-Schritte und ein Modell zu einem Objekt. |
| **ColumnTransformer** | ColumnTransformer | Ein **ColumnTransformer** wendet auf verschiedene Spalten verschiedene Vorbereitungs-Schritte an. |
| **Auffüllen** | imputing | **Auffüllen** ersetzt **fehlende Werte**, zum Beispiel durch den Median der Spalte. |
| **Skalieren** | scaling | **Skalieren** bringt jede Zahlen-Spalte auf Mittelwert 0 und Standard-Abweichung 1. |
| **One-Hot-Kodierung** | one-hot encoding | **One-Hot-Kodierung** macht aus einer Kategorie-Spalte eine Spalte mit 0 und 1 pro Kategorie. |
| **Regularisierung** | regularisation | **Regularisierung** hält die Koeffizienten eines linearen Modells klein, damit das Modell einfach bleibt. |
| **Ridge** | Ridge | **Ridge** ist eine lineare **Regression** mit **Regularisierung**, und `alpha` legt ihre Stärke fest. |
| **Hyper-Parameter** | hyperparameter | Ein **Hyper-Parameter** ist eine Einstellung, die du vor dem Training wählst, zum Beispiel `alpha` von **Ridge**. |
| **Polynom-Merkmale** | polynomial features | **Polynom-Merkmale** ergänzen die Quadrate und Produkte der Zahlen-Spalten. |
| **Kreuz-Validierung** | cross-validation | **Kreuz-Validierung** teilt die **Trainings-Daten** in **Folds** und bewertet das Modell nacheinander auf jedem **Fold**. |
| **Fold** | fold | Ein **Fold** ist einer der gleich großen Teile der **Trainings-Daten** bei der **Kreuz-Validierung**. |
| **Zufalls-Suche** | random search | Eine **Zufalls-Suche** probiert zufällige Werte für **Hyper-Parameter** aus, hier mit `RandomizedSearchCV`. |
| **Gitter-Suche** | grid search | Eine **Gitter-Suche** probiert jeden Wert aus einer festen Liste aus, hier mit `GridSearchCV`. |
| **Über-Anpassung** | overfitting | **Über-Anpassung** heißt, dass das Modell die **Trainings-Daten** viel besser trifft als neue Daten. |
| **Residuum** | residual | Ein **Residuum** ist der echte Preis minus der vorhergesagte Preis. |

### 💻 Wichtiger Code

```python
cat_cols = ["zipcode"]
num_cols = [c for c in X_train.columns if c not in cat_cols]

# Zahlen-Spalten: Median einsetzen und skalieren
num_poly = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("poly", PolynomialFeatures(degree=2, include_bias=False)),
    ("scale", StandardScaler()),
])
# zipcode: häufigsten Wert einsetzen und One-Hot-Kodierung
cat_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])
preprocess_poly = ColumnTransformer([("num", num_poly, num_cols), ("cat", cat_pipe, cat_cols)])
ridge_poly = Pipeline([("preprocess", preprocess_poly), ("model", Ridge())])

# Kreuz-Validierung: erst breit zufällig, dann eng im Gitter
cv = KFold(n_splits=5, shuffle=True, random_state=42)
search = RandomizedSearchCV(ridge_poly, {"model__alpha": loguniform(1e-4, 1e4)}, n_iter=30,
                            cv=cv, scoring="neg_root_mean_squared_error", random_state=42).fit(X_train, y_train)
best = search.best_params_["model__alpha"]
grid = GridSearchCV(ridge_poly, {"model__alpha": np.geomspace(best / 5, best * 5, 11)},
                    cv=cv, scoring="neg_root_mean_squared_error").fit(X_train, y_train)
best_poly = grid.best_estimator_

# Ganz am Ende: Test-Daten mit derselben Funktion vorbereiten und bewerten
test = prepare(df_test)
X_test, y_test = test.drop(columns="price"), test["price"]
residuals = y_test - best_poly.predict(X_test)
```

$$e_i = y_i - \hat{y}_i$$

In Worten: Ein **Residuum** ist der echte Preis minus der vorhergesagte Preis.

Ergebnis in Dollar, die **Ridge**-Werte auf Hunderter gerundet:

| Modell | **RMSE** Training | **RMSE** Test | **R²** Test |
| :--- | ---: | ---: | ---: |
| **Mittelwert-Basis-Modell** | 368.958 | 360.907 | 0,00 |
| **Regel-Basis-Modell** | 267.068 | 260.103 | 0,48 |
| **Ridge** | ≈ 160.200 | ≈ 161.700 | 0,80 |
| **Ridge** mit **Polynom-Merkmalen** | 125.303 | 123.490 | 0,88 |

> 4. Vergleiche am Ende alle Modelle in einer Tabelle mit Trainings-Werten und Test-Werten.
> 5. Die **Residuen** zeigen, dass teure Häuser am schwersten vorherzusagen sind.

## Spick-Zettel: der ganze Ablauf

```python
# 1. Laden und verknüpfen
sales = pd.read_csv("../data/king_county_house_sales.csv")
details = pd.read_csv("../data/king_county_house_details.csv")
assert details["id"].is_unique
df = sales.merge(details, left_on="house_id", right_on="id", how="left", suffixes=("", "_details"))

# 2. Teilen, bevor du untersuchst
df_train, df_test = train_test_split(df, test_size=0.2, random_state=42)

# 3. Änderungen von Hand in einer Funktion
def prepare(data):
    data = data.copy()
    data["age_at_sale"] = pd.to_datetime(data["date"]).dt.year - data["yr_built"]
    data["renovated"] = (data["yr_renovated"] > 0).astype(int)
    return data.drop(columns=["id", "house_id", "id_details", "date", "yr_renovated", "sqft_basement"])

train, test = prepare(df_train), prepare(df_test)
X_train, y_train = train.drop(columns="price"), train["price"]
X_test, y_test = test.drop(columns="price"), test["price"]

# 4. Basis-Modelle
mean_baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
price_per_sqft = (y_train / X_train["sqft_living"]).median()

# 5. Pipeline: Auffüllen, Polynom-Merkmale, Skalieren, One-Hot-Kodierung, Ridge
num_cols = [c for c in X_train.columns if c != "zipcode"]
preprocess = ColumnTransformer([
    ("num", Pipeline([("impute", SimpleImputer(strategy="median")),
                      ("poly", PolynomialFeatures(degree=2, include_bias=False)),
                      ("scale", StandardScaler())]), num_cols),
    ("cat", Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                      ("onehot", OneHotEncoder(handle_unknown="ignore"))]), ["zipcode"]),
])
model = Pipeline([("preprocess", preprocess), ("model", Ridge())])

# 6. Abstimmen mit Kreuz-Validierung
cv = KFold(n_splits=5, shuffle=True, random_state=42)
search = RandomizedSearchCV(model, {"model__alpha": loguniform(1e-4, 1e4)}, n_iter=30, cv=cv,
                            scoring="neg_root_mean_squared_error", random_state=42).fit(X_train, y_train)
best = search.best_params_["model__alpha"]
grid = GridSearchCV(model, {"model__alpha": np.geomspace(best / 5, best * 5, 11)}, cv=cv,
                    scoring="neg_root_mean_squared_error").fit(X_train, y_train)

# 7. Ganz am Ende: Test-Daten bewerten und Residuen ansehen
pred = grid.best_estimator_.predict(X_test)
print(root_mean_squared_error(y_test, pred), mean_absolute_error(y_test, pred), r2_score(y_test, pred))
residuals = y_test - pred
```
